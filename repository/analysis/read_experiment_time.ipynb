# Read experiment wall-clock time

Reads `start_time` / `end_time` / `duration_seconds` from the `.h5` metadata.
Edit the **FILES** list in cell 2 and run all. List several files to compare
(e.g. a `load_atom` run vs a `load_atom_deterministic` run).

Note: these fields only exist on runs saved after the timing change. Older
files show `duration: n/a`.

```python
import sys, os, datetime
sys.path.insert(0, r"C:\Networking Experiment\Artiq7_TEST_NEW\useful_codes\repository")
from analysis.data_loader import load
print("ready")
```

```python
# EDIT: one or more .h5 result files.
FILES = [
    r"C:\Networking Experiment\Artiq7_TEST_NEW\useful_codes\repository\results\2026-05-22\09\000000445-RamseyQaxisScan.h5"
]
```

```python
def _s(v):
    return v.decode() if isinstance(v, bytes) else v

def fmt_hms(sec):
    s = int(round(sec))
    h, rem = divmod(s, 3600)
    m, s = divmod(rem, 60)
    return f"{h}h{m:02d}m{s:02d}s" if h else (f"{m}m{s:02d}s" if m else f"{s}s")

def wall_seconds(meta):
    if meta.get("duration_seconds") is not None:
        return float(meta["duration_seconds"])
    st, et = _s(meta.get("start_time")), _s(meta.get("end_time"))
    if st and et:
        return (datetime.datetime.fromisoformat(et) - datetime.datetime.fromisoformat(st)).total_seconds()
    return None

for f in FILES:
    meta = load(f)["metadata"]
    dur = wall_seconds(meta)
    print(f"RID {meta.get('rid','?')}  {_s(meta.get('experiment','?'))}")
    print(f"  start    : {_s(meta.get('start_time','n/a'))}")
    print(f"  end      : {_s(meta.get('end_time','n/a'))}")
    if dur is None:
        print("  duration : n/a (file predates timing fields)")
    else:
        print(f"  duration : {dur:.1f} s  ({fmt_hms(dur)})")
    print()
```
